# The rhythm of a college town

## Introduction

Daily life in Berkeley follows several clocks: the hour of the day, the day of the week, and the academic calendar of a large university. This report asks when and where problems arise in the city, and whether police enforcement follows them. It uses four open datasets from the Berkeley Police Department (BPD): calls for service, recorded crime cases, traffic collisions, and vehicle stops. Calls, crime cases, and collisions describe demand, showing what residents report and what police record as having happened. Vehicle stops describe a response, showing actions that officers start themselves.

The analysis has four parts. A context section follows stops and collisions over the years, and a time section compares the hours and weekdays of noise complaints, assaults, collisions, and stops. An academic calendar section then compares semester days, breaks, and event days near campus with the rest of the city, and a location section compares police beats and city blocks. Every result is a descriptive summary of recorded events along with their timing or place, and estimating the effects of enforcement falls outside the scope of this report.

## Data

```python
# Read and format saved results only; the report does no computing
import pandas as pd

from bpdtools.paths import repo_path

# Show long text cells in full
pd.set_option("display.max_colwidth", None)

# Summary of the four datasets, saved by the data overview notebook
results = repo_path("results")
summary = pd.read_csv(results / "01_dataset_summary.csv")

# What one record stands for in each dataset
record_meanings = {
    "Calls for Service": "a call to BPD, or an activity that an officer starts",
    "Crime cases": "a case that BPD recorded",
    "Collisions": "a traffic collision",
    "Vehicle stops": "a stop of a vehicle by an officer",
}

# One row per dataset: meaning, record count, first and last local date
date_range = summary["first_local_date"] + " to " + summary["last_local_date"]
data_table = pd.DataFrame(
    {
        "Dataset": summary["dataset"],
        "One record is": summary["dataset"].map(record_meanings),
        "Records": summary["rows"].map("{:,}".format),
        "Local dates": date_range,
    }
)
data_table.set_index("Dataset")
```

Calls for service and crime cases were limited to 2021–2025 when they were prepared, and this five-year window serves as the analysis window for every comparison that uses them. Although collisions reach back to 2007, the analyses use 2021–2025 or a shorter period. Vehicle stops begin in January 2024, so every comparison that involves stops uses 2024–2025.

### How the records are produced

The four datasets come from different BPD processes, and each process shapes what the records can show. The analyses rest on the assumptions below:

- **Calls as reports.** A call records that someone reported something to BPD, which may differ from what actually happened. Each call for service is a record in the BPD computer-aided dispatch system, created when a person phones, dials 911, or reports online, or when an officer starts an activity such as a traffic stop. About a quarter of all calls (**24.7%**) are officer-initiated, so comparisons between residents' calls and stops exclude officer-initiated calls from the call side.
- **Crime cases as police records.** Each crime case reflects what police recorded. Because the published crime file lists each statute of a case on its own row, crime cases were reduced to one record per case number. Each case carries the time when the incident is believed to have happened. A total of 2,134 crime cases (4.2%) have a time of exactly 00:00:00 and 1,624 crime cases (3.2%) have a time of exactly 12:00:00, about three times as many as at 11:00 (508) or 13:00 (558); both times most likely stand for an unknown time.
- **Early-morning collision times.** Early-morning collision times appear unreliable. Zero collisions are recorded between 00:00 and 00:59, and **48%** of the records between 01:00 and 05:59 have minutes under 10, compared with 20% between 06:00 and 19:59. Every hourly collision result is therefore repeated for 06:00–23:59.
- **Stops and police deployment.** A vehicle stop reflects police deployment, because stop counts depend on where and when officers work as well as on how people drive. Because the stop data list each person in a stop on its own row, stops were reduced to one record per stop.
- **Location detail.** Locations differ in detail across the datasets. Calls carry only a block address and a ZIP Code, whereas crime cases, collisions, and stops carry map points.
- **Local time.** Every time in the analysis is Berkeley local time, including the hours around the daylight saving time changes.

None of the datasets is a complete count of what happened in the city. [Data overview notebook](notebooks/01_data_overview.ipynb)

## Context

### How did vehicle stops and collisions change from 2024 to 2025?

Vehicle stops begin in January 2024, so this section compares 2024 with 2025. Each stop counts once, and the 101 collisions without a date are left out because they cannot be placed in a month.

| Records | 2024 | 2025 | Change |
| --- | ---: | ---: | ---: |
| Vehicle stops | 3,232 | 3,078 | -4.8% |
| Collisions | 877 | 784 | -10.6% |

Both counts fell from 2024 to 2025. Collisions fell by **10.6%**, about twice as much as vehicle stops (**4.8%**).

### Do stops and collisions move together month by month?

![Vehicle stops and collisions per month, 2024-2025](figures/02_a1_monthly_trends.png)

The two series peak in different months. Stops were highest in April and September 2024 (388 and 405), whereas collisions were highest in August and October 2024 (92 and 97). Across the 24 months, the correlation between monthly stops and monthly collisions is close to zero (Pearson r = **-0.03**, 95% interval -0.43 to 0.38). Two checks agree: the correlation of the ranks is -0.05, and the correlation of the month-to-month changes, which removes a slow drift that both series could share, is -0.12. Months with more stops did not tend to have more or fewer collisions.

Two years of stop data give only 24 monthly pairs, so the 95% interval runs from -0.43 to 0.38, and a moderate link in either direction is not ruled out. That interval also assumes the monthly pairs are independent; if neighboring months depend on each other, its coverage could be off. Monthly totals also say nothing about place; the Location section compares collisions and stops by police beat. No month in 2024–2025 has a count of 0, and in a later snapshot of the data a month with 0 would mean no records in the dataset that month rather than no stops or collisions. A longer stop series would narrow the interval and show whether the drop from 2024 to 2025 continues. [Trends notebook](notebooks/02_trends.ipynb)

## Time

### When do noise complaints, assaults, and collisions peak?

Noise calls belong to the night and the weekend: of the 12,693 noise calls in 2021–2025, **56%** came between 22:00 and 01:59, and the busiest hour of the week was Friday at 23:00. Recorded assault cases and collisions occur mostly during the working day. Collisions peak at 17:00, whereas assault cases are spread over the afternoon (the check in the next paragraph explains why their apparent noon peak does not hold). Both assault cases and collisions are most frequent on Fridays.

![Noise calls, assault cases, and collisions by weekday and hour, 2021-2025](figures/03_a2_heatmaps.png)

```python
# Peak hour and busiest weekday of each event type, saved by the clock notebook
peaks = pd.read_csv(results / "03_a2_peaks.csv")

# Keep the three event types; the other rows are the checks described below
main_events = ["Noise calls", "Assault cases", "Collisions"]
main_peaks = peaks[peaks["events"].isin(main_events)]

# Counts with commas, hours as HH:00, and shares as percentages
peak_table = pd.DataFrame(
    {
        "Event": main_peaks["events"],
        "Records": main_peaks["n"].map("{:,}".format),
        "Peak hour": main_peaks["peak_hour"].map("{:02d}:00".format),
        "Share in peak hour": main_peaks["peak_hour_share"].map("{:.1%}".format),
        "Busiest weekday": main_peaks["busiest_weekday"],
    }
)
peak_table.set_index("Event")
```

Three checks address the time problems described in the Data section. In the table, assault cases appear to peak at 12:00, but 65 of the 264 assault cases in that hour have a time of exactly 12:00:00. Removing the 41 assault cases at exactly 00:00:00 alone leaves the peak at 12:00 (7.5% of 3,526 assault cases). Removing both the 41 assault cases at 00:00:00 and the 65 assault cases at 12:00:00 leaves 3,461 assault cases without a clear peak: every hour from 13:00 to 17:00 holds between 6.1% and 6.9% of them, and 17:00 is the highest at 6.9%. Restricting collisions to 06:00–23:59 leaves their peak at 17:00 (8.8% of 3,999 collisions).

### Do calls and stops happen at the same hours?

The comparison between calls and stops uses the overlap of two hourly profiles. An hourly profile gives the share of a group's records in each of the 24 hours. The overlap sums, across all 24 hours, the smaller of the two shares in each hour; it equals 1 when two groups have the same timing and 0 when they have zero hours in common.

![Hourly profiles of residents' calls and officer-initiated stops, 2021-2025](figures/03_a6_hourly_profiles.png)

```python
# Overlap of the hourly profiles of calls and stops, saved by the clock notebook
call_stop_overlaps = pd.read_csv(results / "03_a6_overlap.csv")

# Keep the pairs of a call group with a stop type; the last row of the file
# compares the two sources of traffic stops instead
stop_types = ["Traffic stops", "Pedestrian stops"]
call_groups = ["Noise calls", "Battery calls", "Reckless-driving calls", "Other calls"]
pairs = call_stop_overlaps[call_stop_overlaps["second"].isin(stop_types)]

# One row per call group and one column per stop type, with two decimals
overlap_table = pairs.pivot(index="first", columns="second", values="overlap")
overlap_table = overlap_table.loc[call_groups, stop_types].map("{:.2f}".format)
overlap_table.columns = ["Overlap with traffic stops", "Overlap with pedestrian stops"]
overlap_table.rename_axis("Residents' calls")
```

Traffic stops are spread across the day and the night, with their highest hour at 22:00. They share the least timing with noise calls (0.60), which crowd into a few late-night hours. Pedestrian stops concentrate at midday, with 55% happening between 11:00 and 16:59; their hourly profile is close to that of battery calls (0.85) and far from that of noise calls (0.39). In 2024–2025, the traffic stops logged as calls for service and the stops in the separate stop data have almost the same hourly profile (overlap 0.94), so both sources describe the same timing.

To check whether these timing differences could happen by chance, a permutation test shuffled the group labels 9,999 times (using a fixed random seed for reproducibility) for each pairing of a named call type (noise, battery, and reckless driving) with each stop type, while the large group of other calls is summarized by its overlap alone. Every test gave p = 0.0001, the smallest value that 9,999 shuffles allow, meaning differences this large are very unlikely under random chance. With thousands of records in each group, the test detects even small differences, and it treats records as independent of one another. The overlap is therefore the measure to read. For each of the six pairs, none of the 9,999 shuffles produced hourly patterns with as little overlap as we observed. The overlap scores show how large those differences were.

### Do traffic stops follow collisions?

![Hourly profiles of vehicle stops and collisions on weekdays and weekends, 2024-2025](figures/03_a7_stops_collisions.png)

```python
# Overlap of vehicle stops and collisions, saved by the clock notebook
day_overlaps = pd.read_csv(results / "03_a7_overlap.csv")

# Labels of the two hour ranges in the file
ALL_HOURS = "00:00-23:59"
DAYTIME_HOURS = "06:00-23:59"

# Overlap for both hour ranges side by side, one row per kind of day
by_hours = day_overlaps.pivot(index="days", columns="hours", values="overlap")

# Record counts of the comparison over all hours
counts = day_overlaps[day_overlaps["hours"] == ALL_HOURS].set_index("days")

# Overlaps with two decimals and counts with commas
day_table = pd.DataFrame(
    {
        "Overlap, all hours": by_hours[ALL_HOURS].map("{:.2f}".format),
        "Overlap, 06:00–23:59": by_hours[DAYTIME_HOURS].map("{:.2f}".format),
        "Stops": counts["n_stops"].map("{:,}".format),
        "Collisions": counts["n_collisions"].map("{:,}".format),
    }
)
day_table.loc[["All days", "Weekdays", "Weekends"]].rename_axis("Days")
```

On weekdays, vehicle stops peak at 08:00 (7.5% of weekday stops), while collisions peak at 17:00 (10.4% of weekday collisions), yet the two hourly profiles still share most of their timing (overlap **0.81**). On weekends the overlap falls to 0.72: 46% of weekend stops happen between 19:00 and 00:59, whereas weekend collisions peak at 16:00. Leaving out the hours before 06:00 raises both overlaps while keeping the gap between weekdays (0.84) and weekends (0.78). Traffic enforcement follows the timing of collisions more closely on weekdays than on weekends. [Time-of-day notebook](notebooks/03_clock.ipynb)

## Academic calendar

### Did the semester bring a larger change near campus?

In Fall 2024, noise calls in Southside averaged **3.73 per day during the semester and 0.71 during winter break**. The semester rate was 5.27 times the break rate there and 1.88 times the break rate in the other mapped Berkeley police beats. The comparison uses 122 semester days and 24 break days, including covered days with zero events. Southside means beats 6 and 7, immediately south of campus. The city beat map excludes campus itself.

```python
# Select the primary Southside definition and the two outcomes in the report.
rates = pd.read_csv(repo_path("data/processed/a3_period_rates.csv"))
southside = (
    rates.loc[rates["definition"].eq("southside")]
    .set_index(["area", "period", "outcome"])
    .sort_index()
)
outcome_names = {"noise": "Noise calls", "assault": "Assault cases"}
semester = southside.loc[("near", "semester"), :]
winter = southside.loc[("near", "winter_break"), :]
other = southside.loc[("other", "semester"), :]
southside_table = pd.DataFrame(
    {
        "Southside, semester": semester["events"],
        "Southside, break": winter["events"],
        "Southside rate ratio": semester["semester_to_break_ratio"].map(
            "{:.2f}".format
        ),
        "Other beats’ rate ratio": other["semester_to_break_ratio"].map(
            "{:.2f}".format
        ),
    }
)
southside_table = southside_table.loc[list(outcome_names)].rename(index=outcome_names)
southside_table.rename_axis("Recorded outcome")
```

The counts in the middle columns are totals. Each ratio divides semester events per covered day by break events per covered day. Adding beats 1 and 4 to the near-campus group preserved the larger semester increase in noise calls (4.39 versus 1.72 elsewhere). For recorded assaults, expanding the boundary reversed the ordering (1.00 versus 1.21). A stronger semester pattern near campus was clear for noise and depended on the boundary for assault. The geographic comparison excluded 13 noise calls and 10 assault cases with no beat during the semester. All break records for both outcomes had a beat. [Rate table](data/processed/a3_period_rates.csv) · [Geographic exclusions](data/processed/a3_unassigned.csv)

### What happened on event days?

Across Berkeley, the seven home football dates averaged about three more noise calls per day than comparable ordinary dates. Recorded assaults averaged fewer on game days. Halloween had the largest noise difference in the table, but that comparison rests on a single Thursday.

```python
# Display the saved event and matched-day means side by side.
summary = pd.read_csv(repo_path("results/04_event_summary.csv")).set_index("event")
event_names = {
    "home_football": "Home football",
    "move_in": "New-student move-in",
    "halloween": "Halloween",
    "finals": "Finals",
}
event_table = pd.DataFrame(index=summary.index)
event_table["Dates with a match"] = (
    summary["matched_days"].astype(str) + " of " + summary["event_days"].astype(str)
)
for outcome, label in [
    ("noise_calls", "Noise calls"),
    ("assault_cases", "Recorded assault cases"),
    ("alcohol_collisions", "Alcohol-flagged collisions"),
]:
    event_table[label] = (
        summary[f"event_{outcome}"].map("{:.2f}".format)
        + " / "
        + summary[f"mean_control_{outcome}"].map("{:.2f}".format)
    )
event_table.loc[list(event_names)].rename(index=event_names).rename_axis(
    "Fall 2024 event"
)
```

Each pair gives the **event-day mean / matched ordinary-day mean**. Each event date gets equal weight. Its comparison is the average on eligible ordinary dates of the same weekday, month and year. August 21 and December 20 had no eligible comparison and drop out of both sides. Some games share an ordinary Saturday. Only two game-day collisions had a primary factor explicitly mentioning alcohol or a blood alcohol threshold, so an additional record would noticeably change that average. ## Location

### Where did reckless-driving calls line up with collisions and stops?

Among Berkeley’s 14 police beats in 2024–2025, beats with more reckless-driving calls also tended to have more recorded collisions (Spearman ρ = **0.753**). The association with vehicle-stop counts was weaker (**0.191**). Each beat contributes one pair of counts to each comparison. Of 1,091 eligible calls, 1,051 could be assigned to a beat.

The boundary check shows which spatial conclusions hold up. Of 59 collisions within one centimetre of two beat boundaries, 58 could be reassigned to the other beat. Ten beats changed collision rank, yet the calls–collisions correlation was still 0.730. The broad association was more stable than the rank of an individual beat. [Driving-by-beat analysis](notebooks/06_driving_places.ipynb)

### Are noise calls concentrated on a few blocks?

The 12,693 noise calls of 2021–2025 fall on 1,626 block addresses and intersections, and a small share of these places holds a large share of the calls. The busiest 1% of blocks (16) hold **18.9%** of the noise calls, the busiest 5% (81) hold **44.3%**, and the busiest 10% (163) hold **59.4%**. The busiest block, the 2700 block of Channing Way, had 215 noise calls.

![Share of noise calls held by the busiest blocks, 2021-2025](figures/05_a8_concentration.png)

| Busiest blocks | Blocks | Noise calls | Share of all noise calls |
| --- | ---: | ---: | ---: |
| Top 1% | 16 | 2,403 | 18.9% |
| Top 5% | 81 | 5,618 | 44.3% |
| Top 10% | 163 | 7,544 | 59.4% |

The fifth-busiest block, 2100 Martin Luther King Jr Way, is the police headquarters, and its 185 calls may have been logged at the station's address rather than made about that block. Without the headquarters block, without the 588 intersections, or without both, the busiest 5% of blocks still hold between 41.3% and 43.7% of the noise calls. [Noise hotspots analysis](notebooks/05_noise_hotspots.ipynb)

### Do the same blocks stay noise hotspots?

Many hotspots stay hotspots. From one year to the next, between **54%** and **70%** of a year's top 20 blocks were also in the next year's top 20. With a top 10 instead, the overlap was 30% to 60%, and with a top 50 it was 50% to 63%. Seven blocks were in the top 20 in all five years: the police headquarters block, five blocks in Southside (beats 6 and 7), and the 2700 block of Bancroft Way at the edge of campus, outside the beat map. Without the headquarters block or the intersections, seven blocks still stayed in the top 20 every year. Other blocks rose for a single year: the 2000 block of Shattuck Avenue had 82 noise calls in 2022 but 3 in 2021 and 3 in 2025.

![Noise calls per year on the 20 busiest blocks, 2021-2025](figures/05_a8_top_blocks_by_year.png)

A few limits apply to these counts. A block address covers a whole block and an intersection covers a corner, so a hotspot is a stretch of street rather than a single building. A call is a report, and one resident can call about the same party several times, so a busy block has many reports, not necessarily many noisy events. A block with no calls in a year has no records in the dataset for that year, which is not the same as a quiet year.

## Discussion

A complaint, a recorded case and a vehicle stop each describe a different interaction with BPD. Noise calls showed the clearest academic-calendar pattern in these data. Reporting habits and the number of people present could both change between semester and winter break. Eight Southside assault cases during break also give that rate ratio a small denominator. Observing several fall semesters and measuring the population present would help distinguish a recurring calendar pattern from the circumstances of this particular winter.

The beat comparison suggests where complaints and collisions overlap enough to motivate a closer look at individual streets. Busy corridors can generate more of both simply because more vehicles travel there. Vehicle miles traveled by beat would allow a comparison of collisions per mile. The current data support comparisons of recorded counts and leave open whether enforcement changes subsequent collision risk.

## Conclusion

The kinds of problems in the data follow different clocks. Noise calls belong to nights and weekends: **56%** in 2021–2025 came between 22:00 and 01:59, peaking Friday at 23:00. Near campus, they follow the academic calendar: in Southside, the beats south of campus, noise calls averaged 3.73 per day in the Fall 2024 semester and 0.71 in winter break, a semester rate 5.27 times the break rate, compared with 1.88 times in the other beats. Across Berkeley, the seven home football dates averaged about three more noise calls per day than comparable ordinary dates. Recorded assault cases and collisions occur mostly during the working day: collisions peak at 17:00, and assault cases spread over the afternoon once times of exactly 00:00:00 or 12:00:00 are set aside.

By hour, traffic stops share the least timing with noise calls (overlap 0.60) and more with battery (0.81) and reckless-driving calls (0.75). Vehicle stops share more timing with collisions on weekdays (0.81) than on weekends (0.72), when 46% of stops happen between 19:00 and 00:59 and collisions peak in the afternoon. Across Berkeley's 14 police beats in 2024–2025, beats with more reckless-driving calls tended to have more collisions (Spearman ρ = 0.753), and the association between these calls and vehicle stops was weak (0.191). Overall, stops line up more closely with daytime calls and weekday collisions than with late-night noise, and places with many stops line up weakly with places residents report for reckless driving. These results describe recorded events and their timing and place.
