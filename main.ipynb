# The rhythm of a college town

## Introduction

Daily life in Berkeley follows several clocks: the hour of the day, the day of the week, and the academic calendar of a large university. This report asks when and where problems arise in the city, and whether police enforcement follows them. It uses four open datasets from the Berkeley Police Department (BPD): calls for service, recorded crime cases, traffic collisions, and vehicle stops. Calls, crime cases, and collisions describe demand, showing what residents report and what police record as having happened. Vehicle stops describe a response, showing actions that officers start themselves.

The analysis has four parts. A context section follows stops and collisions over the years, and a time section compares the hours and weekdays of noise complaints, assaults, collisions, and stops. An academic calendar section then compares semester days, breaks, and event days near campus with the rest of the city, and a location section compares police beats and city blocks. Every result is a descriptive summary of recorded events along with their timing or place, and estimating the effects of enforcement falls outside the scope of this report.

### How the records are produced

The four datasets come from different BPD processes, and each process shapes what the records can show. The analyses rest on the assumptions below:

- **Calls as reports.** A call records that someone reported something to BPD, which may differ from what actually happened. Each call for service is a record in the BPD computer-aided dispatch system, created when a person phones, dials 911, or reports online, or when an officer starts an activity such as a traffic stop. About a quarter of all calls (**24.7%**) are officer-initiated, so comparisons between residents' calls and stops exclude officer-initiated calls from the call side.
- **Crime cases as police records.** Each crime case reflects what police recorded. Because the published crime file lists each statute of a case on its own row, crime cases were reduced to one record per case number. Each case carries the time when the incident is believed to have happened. A total of 2,134 crime cases (4.2%) have a time of exactly 00:00:00 and 1,624 crime cases (3.2%) have a time of exactly 12:00:00, about three times as many as at 11:00 (508) or 13:00 (558); both times most likely stand for an unknown time.
- **Early-morning collision times.** Early-morning collision times appear unreliable. Zero collisions are recorded between 00:00 and 00:59, and **48%** of the records between 01:00 and 05:59 have minutes under 10, compared with 20% between 06:00 and 19:59. Every hourly collision result is therefore repeated for 06:00–23:59.
- **Stops and police deployment.** A vehicle stop reflects police deployment, because stop counts depend on where and when officers work as well as on how people drive. Because the stop data list each person in a stop on its own row, stops were reduced to one record per stop.
- **Location detail.** Locations differ in detail across the datasets. Calls carry only a block address and a ZIP Code, whereas crime cases, collisions, and stops carry map points.
- **Local time.** Every time in the analysis is Berkeley local time, including the hours around the daylight saving time changes.

None of the datasets is a complete count of what happened in the city. [Data overview notebook](notebooks/01_data_overview.ipynb)

## Time

### When do noise complaints, assaults, and collisions peak?

Noise calls belong to the night and the weekend: of the 12,693 noise calls in 2021–2025, **56%** came between 22:00 and 01:59, and the busiest hour of the week was Friday at 23:00. Recorded assault cases and collisions occur mostly during the working day. Collisions peak at 17:00, whereas assault cases are spread over the afternoon (the check in the next paragraph explains why their apparent noon peak does not hold). Both assault cases and collisions are most frequent on Fridays.

![Noise calls, assault cases, and collisions by weekday and hour, 2021-2025](figures/03_a2_heatmaps.png)

| Event | Records | Peak hour | Share in peak hour | Busiest weekday |
| --- | ---: | ---: | ---: | --- |
| Noise calls | 12,693 | 23:00 | 17.2% | Saturday |
| Assault cases | 3,567 | 12:00 | 7.4% | Friday |
| Collisions | 4,213 | 17:00 | 8.3% | Friday |

Three checks address the time problems described in the Data section. In the table, assault cases appear to peak at 12:00, but 65 of the 264 assault cases in that hour have a time of exactly 12:00:00. Removing the 41 assault cases at exactly 00:00:00 alone leaves the peak at 12:00 (7.5% of 3,526 assault cases). Removing both the 41 assault cases at 00:00:00 and the 65 assault cases at 12:00:00 leaves 3,461 assault cases without a clear peak: every hour from 13:00 to 17:00 holds between 6.1% and 6.9% of them, and 17:00 is the highest at 6.9%. Restricting collisions to 06:00–23:59 leaves their peak at 17:00 (8.8% of 3,999 collisions).

### Do calls and stops happen at the same hours?

The comparison between calls and stops uses the overlap of two hourly profiles. An hourly profile gives the share of a group's records in each of the 24 hours. The overlap sums, across all 24 hours, the smaller of the two shares in each hour; it equals 1 when two groups have the same timing and 0 when they have zero hours in common.

![Hourly profiles of residents' calls and officer-initiated stops, 2021-2025](figures/03_a6_hourly_profiles.png)

| Residents' calls | Overlap with traffic stops | Overlap with pedestrian stops |
| --- | ---: | ---: |
| Noise calls | 0.60 | 0.39 |
| Battery calls | 0.81 | 0.85 |
| Reckless-driving calls | 0.75 | 0.80 |
| Other calls | 0.84 | 0.80 |

Traffic stops are spread across the day and the night, with their highest hour at 22:00. They share the least timing with noise calls (0.60), which crowd into a few late-night hours. Pedestrian stops concentrate at midday, with 55% happening between 11:00 and 16:59; their hourly profile is close to that of battery calls (0.85) and far from that of noise calls (0.39). In 2024–2025, the traffic stops logged as calls for service and the stops in the separate stop data have almost the same hourly profile (overlap 0.94), so both sources describe the same timing.

A permutation test compared each named call type (noise, battery, and reckless driving) with each stop type by shuffling the group labels 9,999 times with a fixed seed, while the large group of other calls is summarized by its overlap alone. Every test gave p = 0.0001, the smallest value that 9,999 shuffles allow. With thousands of records in each group, the test detects even small differences, and it treats records as independent of one another. The overlap is therefore the measure to read, and the p-values only confirm that every pair differs in timing.

### Do traffic stops follow collisions?

![Hourly profiles of vehicle stops and collisions on weekdays and weekends, 2024-2025](figures/03_a7_stops_collisions.png)

| Days | Overlap, all hours | Overlap, 06:00–23:59 | Stops | Collisions |
| --- | ---: | ---: | ---: | ---: |
| All days | 0.81 | 0.85 | 6,310 | 1,661 |
| Weekdays | 0.81 | 0.84 | 4,910 | 1,302 |
| Weekends | 0.72 | 0.78 | 1,400 | 359 |

On weekdays, vehicle stops peak at 08:00 (7.5% of weekday stops), while collisions peak at 17:00 (10.4% of weekday collisions), yet the two hourly profiles still share most of their timing (overlap **0.81**). On weekends the overlap falls to 0.72: 46% of weekend stops happen between 19:00 and 00:59, whereas weekend collisions peak at 16:00. Leaving out the hours before 06:00 raises both overlaps while keeping the gap between weekdays (0.84) and weekends (0.78). Traffic enforcement follows the timing of collisions more closely on weekdays than on weekends. [Time-of-day notebook](notebooks/03_clock.ipynb)

## Academic calendar

### Did the semester bring a larger change near campus?

In Fall 2024, noise calls in Southside averaged **3.73 per day during the semester and 0.71 during winter break**. The semester rate was 5.27 times the break rate there and 1.88 times the break rate in the other mapped Berkeley police beats. The comparison uses 122 semester days and 24 break days, including covered days with zero events. Southside means beats 6 and 7, immediately south of campus. The city beat map excludes campus itself.

| Recorded outcome | Southside, semester | Southside, break | Southside rate ratio | Other beats’ rate ratio |
| --- | ---: | ---: | ---: | ---: |
| Noise calls | 455 | 17 | 5.27 | 1.88 |
| Assault cases | 52 | 8 | 1.28 | 1.09 |

The counts in the middle columns are totals. Each ratio divides semester events per covered day by break events per covered day. Adding beats 1 and 4 to the near-campus group preserved the larger semester increase in noise calls (4.39 versus 1.72 elsewhere). For recorded assaults, expanding the boundary reversed the ordering (1.00 versus 1.21). A stronger semester pattern near campus was clear for noise and depended on the boundary for assault. The geographic comparison excluded 13 noise calls and 10 assault cases with no beat during the semester. All break records for both outcomes had a beat. ### What happened on event days?

Across Berkeley, the seven home football dates averaged about three more noise calls per day than comparable ordinary dates. Recorded assaults averaged fewer on game days. Halloween had the largest noise difference in the table, but that comparison rests on a single Thursday.

| Fall 2024 event | Dates with a match | Noise calls | Recorded assault cases | Alcohol-flagged collisions |
| --- | ---: | ---: | ---: | ---: |
| Home football | 7 of 7 | 19.71 / 17.00 | 1.86 / 3.62 | 0.29 / 0.05 |
| New-student move-in | 1 of 2 | 7 / 6 | 3 / 1 | 0 / 0 |
| Halloween | 1 of 1 | 16 / 6 | 1 / 5 | 0 / 0 |
| Finals | 4 of 5 | 4.00 / 4.25 | 1.75 / 2.25 | 0 / 0 |

Each pair gives the **event-day mean / matched ordinary-day mean**. Each event date gets equal weight. Its comparison is the average on eligible ordinary dates of the same weekday, month and year. August 21 and December 20 had no eligible comparison and drop out of both sides. Some games share an ordinary Saturday. Only two game-day collisions had a primary factor explicitly mentioning alcohol or a blood alcohol threshold, so an additional record would noticeably change that average. ## Location

### Where did reckless-driving calls line up with collisions and stops?

Among Berkeley’s 14 police beats in 2024–2025, beats with more reckless-driving calls also tended to have more recorded collisions (Spearman ρ = **0.753**). The association with vehicle-stop counts was weaker (**0.191**). Each beat contributes one pair of counts to each comparison. Of 1,091 eligible calls, 1,051 could be assigned to a beat.

The boundary check shows which spatial conclusions hold up. Of 59 collisions within one centimetre of two beat boundaries, 58 could be reassigned to the other beat. Ten beats changed collision rank, yet the calls–collisions correlation was still 0.730. The broad association was more stable than the rank of an individual beat. ## Discussion

A complaint, a recorded case and a vehicle stop each describe a different interaction with BPD. Noise calls showed the clearest academic-calendar pattern in these data. Reporting habits and the number of people present could both change between semester and winter break. Eight Southside assault cases during break also give that rate ratio a small denominator. Observing several fall semesters and measuring the population present would help distinguish a recurring calendar pattern from the circumstances of this particular winter.

The beat comparison suggests where complaints and collisions overlap enough to motivate a closer look at individual streets. Busy corridors can generate more of both simply because more vehicles travel there. Vehicle miles traveled by beat would allow a comparison of collisions per mile. The current data support comparisons of recorded counts and leave open whether enforcement changes subsequent collision risk.

## Conclusion

The kinds of problems in the data follow different clocks. Noise calls belong to nights and weekends: **56%** in 2021–2025 came between 22:00 and 01:59, peaking Friday at 23:00. Near campus, they follow the academic calendar: in Southside, the beats south of campus, noise calls averaged 3.73 per day in the Fall 2024 semester and 0.71 in winter break, a semester rate 5.27 times the break rate, compared with 1.88 times in the other beats. Across Berkeley, the seven home football dates averaged about three more noise calls per day than comparable ordinary dates. Recorded assault cases and collisions occur mostly during the working day: collisions peak at 17:00, and assault cases spread over the afternoon once times of exactly 00:00:00 or 12:00:00 are set aside.

By hour, traffic stops share the least timing with noise calls (overlap 0.60) and more with battery (0.81) and reckless-driving calls (0.75). Vehicle stops share more timing with collisions on weekdays (0.81) than on weekends (0.72), when 46% of stops happen between 19:00 and 00:59 and collisions peak in the afternoon. Across Berkeley's 14 police beats in 2024–2025, beats with more reckless-driving calls tended to have more collisions (Spearman ρ = 0.753), and the association between these calls and vehicle stops was weak (0.191). Overall, stops line up more closely with daytime calls and weekday collisions than with late-night noise, and places with many stops line up weakly with places residents report for reckless driving. These results describe recorded events and their timing and place.
