## Personal reflection

### John Hu

When I reassigned collisions along beat boundaries, I expected only a few small shifts. Instead, 10 of the 14 beats changed rank. I was surprised that one coding decision could reshuffle so much of the list. Yet the correlation between reckless-driving calls and collisions barely moved, from 0.753 to 0.730.

That contrast changed how I read my results. Under both ways of assigning boundary collisions, beats with more reckless-driving calls generally had more recorded collisions. I felt comfortable reporting that pattern. I had much less confidence in singling out one beat as the “worst.” When two beats have similar collision counts, moving a few records across their shared boundary can change which one ranks higher.

I also thought about how someone might use that list. A reader could point to a high ranking when deciding where to send patrols or describing a neighborhood as dangerous. Our table counts recorded collisions, and even the order of those counts depends partly on how we handle records at the boundary. Next time, I’ll run that check before deciding which places to highlight. If I publish a ranking, I’ll show the two rankings side by side so readers can see which beats move and by how much.

### Minchee Yang

When I first made the hourly table, assaults seemed to peak at 12:00, and I had already written that into my draft before checking the exact timestamps. It turned out that 65 of the 264 assault cases in that hour were logged at exactly 12:00:00. Across all crime records, 1,624 were recorded at 12:00:00—about three times as many as at 11:00:00 or 13:00:00. Times like 12:00:00 are most likely just default placeholders when the actual time isn't known. After I filtered out those cases along with the 00:00:00 ones, the noon peak disappeared, and every hour from 13:00 to 17:00 ended up having roughly the same share of assaults, between 6.1% and 6.9%. That was honestly pretty frustrating, because the noon spike looked like such a clear result at first, and I still had to go back and rewrite the parts where I had already used it.

Time zones were easily the most confusing part. At first, I assumed the calls-for-service file stored timestamps in UTC, since an older online version of the same dataset did. However, when I checked the first and last rows, it turned out to already use local time with the offset included, whereas the collision table kept a UTC timestamp right next to its local hour. Catching these mismatches definitely changed how I handle timestamps now.

The permutation tests taught me a similar lesson. Every single test gave p = 0.0001—the lowest value possible with 9,999 shuffles—simply because each group had thousands of records. With sample sizes that large, the p-values couldn't tell a big timing difference apart from a small one. So we compared the groups by their overlap, and the tests themselves only showed that no two groups had the exact same timing. Next time, I'll make sure to pick a metric that actually matters before running the test.

### Winny Wen

Before I ran the numbers, I expected stops and collisions to rise and fall together. Both did fall from 2024 to 2025, stops by 4.8% and collisions by 10.6%, so at first it looked like one shared trend. But when I lined up the 24 months, the correlation was only r = −0.03. Stops peaked in April and September 2024, while collisions peaked in August and October. A drop across two years said nothing about whether the two moved together month by month, and I had to be careful not to let the yearly totals suggest a link that the monthly data didn't support.

The most confusing part was deciding what one record means. The stop file lists every person in a stop on its own row, so counting rows would have counted some stops more than once. I counted each stop once by its record ID and wrote tests for repeated IDs and missing dates. I also had to think about what a month with zero would mean. John suggested we describe it as no records in the dataset that month, because a zero could come from gaps in reporting rather than a quiet month. None of our 24 months turned out to be zero, but it made me realize how much a count depends on how the data was recorded.

In A8, one of the busiest "blocks" turned out to be the police headquarters, which reminded me that an address records where a call was logged, not always where the noise happened, so now I check what a location really stands for before calling it a hotspot.

Looking at the project as a whole, what I'll take away most is how much our reviews shaped the results.
