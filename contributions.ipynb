## Personal reflection

### John Hu

When I reassigned collisions along beat boundaries, I expected only a few small shifts. Instead, 10 of the 14 beats changed rank. I was surprised that one coding decision could reshuffle so much of the list. Yet the correlation between reckless-driving calls and collisions barely moved, from 0.753 to 0.730.

That contrast changed how I read my results. Under both ways of assigning boundary collisions, beats with more reckless-driving calls generally had more recorded collisions. I felt comfortable reporting that pattern. I had much less confidence in singling out one beat as the “worst.” When two beats have similar collision counts, moving a few records across their shared boundary can change which one ranks higher.

I also thought about how someone might use that list. A reader could point to a high ranking when deciding where to send patrols or describing a neighborhood as dangerous. Our table counts recorded collisions, and even the order of those counts depends partly on how we handle records at the boundary. Next time, I’ll run that check before deciding which places to highlight. If I publish a ranking, I’ll show the two rankings side by side so readers can see which beats move and by how much.
